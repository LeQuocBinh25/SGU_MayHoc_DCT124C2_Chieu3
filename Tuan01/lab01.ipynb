### Bài 6. Phép toán và nhân ma trận

### Bài 7. Các hàm thống kê trên Tensor
##### Tính trung bình, độ lệch chuẩn và tổng tích lũy theo từng cột.

```python
import torch
a = torch.randn(3, 4)
print(a)
print("\nMean:")
print(torch.mean(a, dim=0))
print("\nStandard deviation:")
print(torch.std(a, dim=0))
print("\nCumulative sum:")
print(torch.cumsum(a, dim=0))

print("\n\nThay dim=0 thanh dim=1")
print("Mean:")
print(torch.mean(a, dim=1))
print("\nStandard deviation:")
print(torch.std(a, dim=1))
print("\nCumulative sum:")
print(torch.cumsum(a, dim=1))
```

##### Nhận xét:<br>dim=0: mean,std tính theo từng cột, cumsum cộng dồn từ trên xuống<br>dim=1: mean,std tính theo từng hàng, cumsum cộng dồn từ trái sang

### Bài 9. Gradient Descent tìm hệ số đa thức
##### Sinh dữ liệu từ đa thức y = x2 + 2x + 3, sau đó dùng NAdam và autograd để tìm lại ba hệ số.

```python
import numpy as np
import torch
polynomial = np.poly1d([1, 2, 3])
N = 20
X = np.random.randn(N, 1) * 5
Y = polynomial(X)
XX = np.hstack([
X * X,
X,
np.ones_like(X)
])
w = torch.randn(3, 1, requires_grad=True)
x = torch.tensor(XX, dtype=torch.float32)
y = torch.tensor(Y, dtype=torch.float32)
optimizer = torch.optim.NAdam([w], lr=0.01)
print("Initial coefficients:")
print(w)
for epoch in range(1000):
    optimizer.zero_grad()
    y_pred = x @ w
    mse = torch.mean(torch.square(y - y_pred))
    mse.backward()
    optimizer.step()
print("Learned coefficients:")
print(w)
```

##### optimizer.zero_grad(): xóa gradient của lần lặp trước <br>mse.backward(): dùng để tự động tính đạo hàm<br>optimizer.step(): được dùng để cập nhật các trọng số của mô hình dựa trên các gradient đã được tính toán trong quá trình lan truyền ngược 

### Bài 10. Giải hệ phương trình bằng Autograd
##### Tối thiểu hóa tổng sai số bình phương để tìm A, B, C, D thỏa bốn phương trình.

```python
import random
import torch
A = torch.tensor(random.random(), requires_grad=True)
B = torch.tensor(random.random(), requires_grad=True)
C = torch.tensor(random.random(), requires_grad=True)
D = torch.tensor(random.random(), requires_grad=True)
EPOCHS = 2000
optimizer = torch.optim.NAdam([A, B, C, D], lr=0.01)
for _ in range(EPOCHS):
    y1 = A + B - 9
    y2 = C - D - 1
    y3 = A + C - 8
    y4 = B - D - 2
    
    sqerr = y1*y1 + y2*y2 + y3*y3 + y4*y4
    
    optimizer.zero_grad()
    sqerr.backward()
    optimizer.step()
print("A =", A)
print("B =", B)
print("C =", C)
print("D =", D)
```

##### Nhận xét: Hệ phương trình có thể có vô số nghiệm vì 4 phương trình nhưng chỉ có 3 phương trình độc lập.

#### Autograd

```python
print("x =", x)
print("y =", y)
print("dy/dx bằng autograd =", x.grad)

manual_gradient = 4 * 2 + 5

print("dy/dx tính thủ công =", manual_gradient)

np.random.seed(42)

polynomial = np.poly1d([2, -3, 5])

N = 20

X = np.random.randn(N, 1) * 5
Y = polynomial(X)

XX = np.hstack([
    X * X,
    X,
    np.ones_like(X)
])

x_train = torch.tensor(XX, dtype=torch.float32)
y_train = torch.tensor(Y, dtype=torch.float32)

w = torch.randn(3, 1, requires_grad=True)

optimizer = torch.optim.NAdam([w], lr=0.01)

EPOCHS = 2000

for epoch in range(EPOCHS):

    # Forward
    y_pred = x_train @ w

    # Loss
    mse = torch.mean((y_train - y_pred) ** 2)

    # Backward
    optimizer.zero_grad()
    mse.backward()


    optimizer.step()
```
